````python
def factorial(n):
    """
    Calculate the factorial of a non-negative integer.

    Parameters:
    n (int): The number to find the factorial of.

    Returns:
    int: The factorial of the number.
    """
    # Input validation
    if not isinstance(n, int):
        raise TypeError("Input must be an integer.")
    if n < 0:
        raise ValueError("Input must be a non-negative integer.")

    # Factorial calculation
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
````

````python
# Test the function with valid inputs
for val in [0, 1, 5, 10]:
    print(f"Factorial of {val} is: {factorial(val)}")

# Test input validation
try:
    factorial(-3)
except ValueError as e:
    print(f"Validation check passed for negative input: {e}")

try:
    factorial(5.5)
except TypeError as e:
    print(f"Validation check passed for non-integer input: {e}")
````

````python
def fibonacci(n):
    """
    Generate a list containing the first n Fibonacci numbers.

    Parameters:
    n (int): The number of Fibonacci numbers to generate.

    Returns:
    list: A list containing the first n Fibonacci numbers.
    """
    # Input validation
    if not isinstance(n, int):
        raise TypeError("Input n must be an integer.")
    if n <= 0:
        raise ValueError("Input n must be a positive integer.")

    # Generate Fibonacci sequence
    sequence = []
    if n >= 1:
        sequence.append(0)
    if n >= 2:
        sequence.append(1)

    while len(sequence) < n:
        next_num = sequence[-1] + sequence[-2]
        sequence.append(next_num)

    return sequence
````

````python
# Test the Fibonacci function with valid inputs
for val in [1, 2, 5, 10]:
    print(f"First {val} Fibonacci numbers: {fibonacci(val)}")

# Test input validation
try:
    fibonacci(0)
except ValueError as e:
    print(f"Validation check passed for non-positive input: {e}")

try:
    fibonacci("5")
except TypeError as e:
    print(f"Validation check passed for non-integer input: {e}")
````

### C Struct to Python Dictionary

The C struct:
```c
struct Student {
    int roll;
    char name[20];
    float marks;
};
```

In Python, this structural representation can be modeled natively as a dictionary where the member names are keys and their values correspond to the respective data types (`int`, `str`, and `float`).

````python
# Representing the struct Student as a Python dictionary
student = {
    "roll": 101,
    "name": "Alice",
    "marks": 92.5
}

# Displaying the dictionary structure and types
for key, value in student.items():
    print(f"{key}: {value} (type: {type(value).__name__})")
````
